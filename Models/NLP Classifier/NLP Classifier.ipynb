```python
import os
import sys
CHUNK_SIZE = 50
```

```python
labels_train = [] # 0 for english and 1 for not english
data_train = []
data_test = []
def load_phrases(file_path):
    num = 0
    labels = []
    data = []
    with open(file_path, "r", encoding="mbcs") as f:
        for section in chunk(f, "\t"):
            if (num != 0):
                parts = section.split("\n", 1)
                data.extend(parts)
                if (len(parts) == 1): # Last phrase
                    labels.append(num % 2)
                else:
                    labels.extend([1, 0])
            else:
                labels.append(num % 2)
                data.append(section)
            num += 1
    return labels, data

# Read and seperate in chunks to prevent memory issues
def chunk(file, seperator):
    buffer = ""
    while True:
        chunk = file.read(CHUNK_SIZE)
        if not chunk:
            if buffer:
                yield buffer
            break
        buffer += chunk
        while True:
            try:
                phrase, buffer = buffer.split(seperator, 1)
            except ValueError:
                break
            else:
                yield phrase

labels_train, data_train = load_phrases("C:/Users/sbodd/Challenge Data/challenge-data/challenge-data/train.txt")
_, data_test = load_phrases("C:/Users/sbodd/Challenge Data/challenge-data/challenge-data/test.rand.txt")
```

```python
import nltk
import sklearn
import numpy as np
nltk.download()
import functools
from functools import reduce
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer
from sklearn.feature_extraction.text import TfidfVectorizer
# All of the above imports are only algorithms (I'm fairly certain that none of them use a corpus or such)

tokens = []
stemmer = PorterStemmer()
for section in data_train:
    words = word_tokenize(section)
    stemmed_sentence = reduce(lambda x, y: x + stemmer.stem(y), words, "")
    tokens.append(stemmed_sentence)

from sklearn.ensemble import RandomForestClassifier
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(tokens)
model = RandomForestClassifier(n_estimators = 350, verbose = 10, n_jobs = -1, max_depth = 280)
model.fit(X, labels_train)

test_tokens = []
for section in data_train:
    words = word_tokenize(section)
    stemmed_sentence = reduce(lambda x, y: x + stemmer.stem(y), words, "")
    test_tokens.append(stemmed_sentence)    
X_test = vectorizer.transform(test_tokens)

prediction = model.predict(X_test)
print("Prediction: ", prediction) # For debugging purposes
print(prediction[4765])
```
